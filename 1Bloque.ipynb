```python

!git clone https://github.com/camiloascencio-hash/GP-S.git

import pandas as pd
import numpy as np
import json



with open("GP-S/datos/metadatos.json", "r", encoding="utf-8") as archivo:
    metadatos = json.load(archivo)

metadatos

def leer(nombre_archivo):
    return pd.read_csv("GP-S/datos/" + nombre_archivo)

llen_raw = leer("calidad_llenado.csv")
atr_raw = leer("calidad_atributos.csv")
spec = leer("especificaciones_calidad.csv")
maestro = leer("maestro_sku.csv")
dem_raw = leer("demanda_semanal.csv")

for nombre, d in [("calidad_llenado", llen_raw), ("calidad_atributos", atr_raw),
                  ("especificaciones", spec), ("maestro_sku", maestro),
                  ("demanda_semanal", dem_raw)]:
    print(nombre, d.shape[0], "filas x", d.shape[1], "columnas")

# los datos crudos no se modifican: se trabaja sobre copias
llen = llen_raw.copy()
atr = atr_raw.copy()
dem = dem_raw.copy()

llen.head()

atr.head()

spec

bitacora = []

def registrar(archivo, chequeo, resultado, decision):
    bitacora.append({"archivo": archivo, "chequeo": chequeo,
                     "resultado": resultado, "decision": decision})

faltantes = llen.isna().sum().sum()
registrar("calidad_llenado", "Valores faltantes", str(faltantes) + " celdas vacías", "Sin acción")

duplicados = llen.duplicated(["subgrupo", "n_obs"]).sum()
registrar("calidad_llenado", "Duplicados (subgrupo, n_obs)", str(duplicados) + " duplicados", "Sin acción")

tam = llen.groupby("subgrupo").size()
tamanos = [int(v) for v in sorted(tam.unique())]
registrar("calidad_llenado", "Tamaño de subgrupo",
          "n = " + str(tamanos) + " en " + str(len(tam)) + " subgrupos",
          "n = 5 constante, se usan las constantes de la carta X-R")

mezclas = llen.groupby("subgrupo")[["linea", "sku", "turno"]].nunique().max().max()
registrar("calidad_llenado", "Subgrupos que mezclan línea, SKU o turno",
          "máximo " + str(mezclas) + " valor distinto por subgrupo",
          "Sin acción, los subgrupos son racionales")

imposibles = (llen["volumen_mL"] <= 0).sum()
registrar("calidad_llenado", "Volúmenes menores o iguales a cero", str(imposibles) + " valores", "Sin acción")

cuenta = llen.groupby(["linea", "formato_L", "turno"])["subgrupo"].nunique()
registrar("calidad_llenado", "Subgrupos por línea, formato y turno",
          "mínimo " + str(cuenta.min()) + ", máximo " + str(cuenta.max()),
          "Se reporta el desbalance (turno B de 1,5 L tiene menos subgrupos)")

faltantes_atr = atr.isna().sum().sum()
registrar("calidad_atributos", "Valores faltantes", str(faltantes_atr) + " celdas vacías", "Sin acción")

duplicados_atr = atr.duplicated().sum()
registrar("calidad_atributos", "Filas duplicadas", str(duplicados_atr) + " duplicados", "Sin acción")

mal = (atr["n_defectuosas"] > atr["n_inspeccionadas"]).sum()
registrar("calidad_atributos", "Defectuosas mayores que inspeccionadas", str(mal) + " filas", "Sin acción")

registrar("calidad_atributos", "Tamaño de inspección n",
          "entre " + str(atr["n_inspeccionadas"].min()) + " y " + str(atr["n_inspeccionadas"].max()),
          "n varía, se usa carta p con límites variables")

sin_def = (atr["causa_principal"] == "Sin defectos").sum()
registrar("calidad_atributos", "Categoría 'Sin defectos'", str(sin_def) + " filas",
          "Se mantienen para calcular p, se excluyen del Pareto")

registrar("especificaciones_calidad", "Valores faltantes", str(spec.isna().sum().sum()) + " celdas vacías", "Sin acción")
registrar("maestro_sku", "Valores faltantes", str(maestro.isna().sum().sum()) + " celdas vacías", "Sin acción")

faltantes_dem = dem.isna().sum().sum()
registrar("demanda_semanal", "Valores faltantes",
          str(faltantes_dem) + " celdas vacías (ventas_cajas y dias_sin_stock)",
          "Se revisan las filas; el tratamiento se define en la Parte B")

duplicados_dem = dem.duplicated().sum()
registrar("demanda_semanal", "Filas duplicadas", str(duplicados_dem) + " duplicados", "Sin acción")

# las filas con datos faltantes en demanda
dem[dem.isna().any(axis=1)]

bit_df = pd.DataFrame(bitacora)
bit_df

### Estado de los datos

#Los archivos de calidad (llenado y atributos) están completos: sin faltantes, sin duplicados y sin valores imposibles. Los 840 subgrupos de llenado tienen n = 5 y cada uno corresponde a una sola línea, SKU y turno.

#Hallazgos a documentar:
#- Los subgrupos por turno están desbalanceados (turno B de 1,5 L tiene 76, contra 104-108 de los otros turnos).
#- En calidad_atributos el tamaño de inspección varía (181 a 260), por lo que corresponde carta p. Hay 2 filas con causa "Sin defectos".
#- demanda_semanal tiene 3 filas con ventas_cajas y dias_sin_stock faltantes. Se tratarán en la Parte B.

g = llen.groupby("subgrupo")

sg = g.agg(fecha=("fecha", "first"), hora=("hora", "first"),
           semana=("semana_registro", "first"), turno=("turno", "first"),
           linea=("linea", "first"), formato=("formato_L", "first"),
           sku=("sku", "first"))

sg["xbar"] = g["volumen_mL"].mean()
sg["R"] = g["volumen_mL"].max() - g["volumen_mL"].min()
sg = sg.reset_index()

# el orden en el tiempo sale de fecha y hora (el número de subgrupo no sirve en L2)
sg["fecha_hora"] = pd.to_datetime(sg["fecha"] + " " + sg["hora"])

sg.head()

# Todos los subgrupos tienen n = 5, por lo que corresponde la carta X-R y no la X-S
A2 = 0.577
D3 = 0
D4 = 2.114
d2 = 2.326

resumen = []

# BLOQUE 3.1 (una sola vez)
esp = spec.set_index("formato_L")

capacidades = []     # capacidad en Fase I estable (global y por turno)
desempeno = []       # desempeño en Fase II (solo referencia)
legales = []         # verificación del límite legal
cifras = []          # cifras de apoyo que se citan en el informe

linea = "L1"
formato = 0.5

c = sg[(sg["linea"] == linea) & (sg["formato"] == formato)]
c = c.sort_values("fecha_hora").reset_index(drop=True)
c["orden"] = c.index + 1

f1 = c[c["semana"] <= 5]      # Fase I: se calculan los límites
f2 = c[c["semana"] > 5]       # Fase II: se juzga contra los límites congelados

len(f1), len(f2)

xbb = f1["xbar"].mean()
Rb = f1["R"].mean()

LCSx = xbb + A2 * Rb
LCIx = xbb - A2 * Rb
LCSr = D4 * Rb

fuera = f1[(f1["R"] > LCSr) | (f1["xbar"] > LCSx) | (f1["xbar"] < LCIx)]
print("Ronda 1:", len(fuera), "puntos fuera")

# se sacan los puntos fuera para recalcular
f1 = f1[(f1["R"] <= LCSr) & (f1["xbar"] <= LCSx) & (f1["xbar"] >= LCIx)]

fuera[["orden", "turno", "sku", "xbar", "R"]]

xbb = f1["xbar"].mean()
Rb = f1["R"].mean()

LCSx = xbb + A2 * Rb
LCIx = xbb - A2 * Rb
LCSr = D4 * Rb

fuera = f1[(f1["R"] > LCSr) | (f1["xbar"] > LCSx) | (f1["xbar"] < LCIx)]
print("Ronda 2:", len(fuera), "puntos fuera")

f1 = f1[(f1["R"] <= LCSr) & (f1["xbar"] <= LCSx) & (f1["xbar"] >= LCIx)]

fuera[["orden", "turno", "sku", "xbar", "R"]]

xbb = f1["xbar"].mean()
Rb = f1["R"].mean()

LCSx = xbb + A2 * Rb
LCIx = xbb - A2 * Rb
LCSr = D4 * Rb

fuera = f1[(f1["R"] > LCSr) | (f1["xbar"] > LCSx) | (f1["xbar"] < LCIx)]
print("Ronda 3:", len(fuera), "puntos fuera")

f1 = f1[(f1["R"] <= LCSr) & (f1["xbar"] <= LCSx) & (f1["xbar"] >= LCIx)]

fuera[["orden", "turno", "sku", "xbar", "R"]]

xbb = f1["xbar"].mean()
Rb = f1["R"].mean()

LCSx = xbb + A2 * Rb
LCIx = xbb - A2 * Rb
LCSr = D4 * Rb
LCIr = D3 * Rb
sigma = Rb / d2

print("x doble barra:", round(xbb, 2), " R barra:", round(Rb, 2), " sigma:", round(sigma, 3))
print("LCS X:", round(LCSx, 2), " LCI X:", round(LCIx, 2), " LCS R:", round(LCSr, 2))

# puntos de Fase I que quedaron fuera del cálculo
fase1_completa = c[c["semana"] <= 5]
excluidos = fase1_completa[fase1_completa["orden"].isin(f1["orden"]) == False]

excluidos[["orden", "fecha_hora", "turno", "sku", "xbar", "R"]]

c["fuera_x"] = (c["xbar"] > LCSx) | (c["xbar"] < LCIx)
c["fuera_r"] = (c["R"] > LCSr) | (c["R"] < LCIr)

f2 = c[c["semana"] > 5]

print("Fase II, puntos fuera en X:", f2["fuera_x"].sum(), "de", len(f2))
print("Fase II, puntos fuera en R:", f2["fuera_r"].sum(), "de", len(f2))

resumen.append([linea, formato, len(fase1_completa), len(excluidos),
                xbb, Rb, LCSx, LCIx, LCSr, sigma,
                f2["fuera_x"].sum(), f2["fuera_r"].sum()])

# BLOQUES 3.2 a 3.6 (capacidad, desempeño, límite legal y cifras de esta combinación)
LIE = esp.loc[formato, "LIE_mL"]
LSE = esp.loc[formato, "LSE_mL"]

d = llen[llen["subgrupo"].isin(f1["subgrupo"])]      # mediciones de la Fase I estable

g = d.groupby("subgrupo")["volumen_mL"]
mu, Rb = g.mean().mean(), (g.max() - g.min()).mean()
sigma_c = Rb / d2                                    # corto plazo -> Cp, Cpk
sigma_l = d["volumen_mL"].std(ddof=1)                # largo plazo -> Pp, Ppk

Cp = (LSE - LIE) / (6 * sigma_c)
Cpk = min(LSE - mu, mu - LIE) / (3 * sigma_c)
Pp = (LSE - LIE) / (6 * sigma_l)
Ppk = min(LSE - mu, mu - LIE) / (3 * sigma_l)

print(linea, formato, "Global")
print("Cp =", round(Cp, 3), " Cpk =", round(Cpk, 3), " Pp =", round(Pp, 3), " Ppk =", round(Ppk, 3))

capacidades.append([linea, formato, "Global", len(f1), mu, sigma_c, sigma_l, Cp, Cpk, Pp, Ppk])

for turno in sorted(d["turno"].unique()):
    d_t = d[d["turno"] == turno]

    g = d_t.groupby("subgrupo")["volumen_mL"]
    mu, Rb = g.mean().mean(), (g.max() - g.min()).mean()
    sigma_c = Rb / d2
    sigma_l = d_t["volumen_mL"].std(ddof=1)

    Cp = (LSE - LIE) / (6 * sigma_c)
    Cpk = min(LSE - mu, mu - LIE) / (3 * sigma_c)
    Pp = (LSE - LIE) / (6 * sigma_l)
    Ppk = min(LSE - mu, mu - LIE) / (3 * sigma_l)

    capacidades.append([linea, formato, turno, g.ngroups, mu, sigma_c, sigma_l, Cp, Cpk, Pp, Ppk])

d_f2 = llen[llen["subgrupo"].isin(f2["subgrupo"])]

mu = d_f2["volumen_mL"].mean()
sigma_l = d_f2["volumen_mL"].std(ddof=1)

Pp = (LSE - LIE) / (6 * sigma_l)
Ppk = min(LSE - mu, mu - LIE) / (3 * sigma_l)

desempeno.append([linea, formato, len(f2), mu, sigma_l, Pp, Ppk])

nominal = esp.loc[formato, "nominal_mL"]
T = esp.loc[formato, "tolerancia_oiml_mL"]

x = llen[(llen["linea"] == linea) & (llen["formato_L"] == formato)]["volumen_mL"]   # todas las mediciones

media = x.mean()
error_T1 = ((x < nominal - T) & (x >= nominal - 2 * T)).sum()    # deficiencia mayor que T y hasta 2T
error_T2 = (x < nominal - 2 * T).sum()                           # deficiencia mayor que 2T
pct_T1 = 100 * error_T1 / len(x)

# R87: (1) media >= nominal, (2) a lo más 2,5 % con error T1, (3) ninguno con error T2
cumple = (media >= nominal) and (pct_T1 <= 2.5) and (error_T2 == 0)

legales.append([linea, formato, nominal, T, len(x), media, x.min(), error_T1, pct_T1, error_T2, cumple])

centro = (LSE + LIE) / 2
k_desc = abs(xbb - centro) / ((LSE - LIE) / 2)       # descentrado (lámina 22): 0 = centrado
sigma_meta = (LSE - LIE) / (6 * 1.33)                # sigma necesaria para llegar a Cp = 1,33
reduccion = 100 * (1 - sigma_meta / sigma)           # % que habría que reducir sigma

sobre_f1 = (d["volumen_mL"] > LSE).sum()             # botellas sobre el LSE, Fase I estable
sobre_f2 = (d_f2["volumen_mL"] > LSE).sum()          # botellas sobre el LSE, Fase II
bajo_lie = (x < LIE).sum()                           # botellas bajo el LIE, todas las mediciones

cifras.append([linea, formato, k_desc, sigma, sigma_meta, reduccion,
               len(d), sobre_f1, len(d_f2), sobre_f2, bajo_lie])

import matplotlib.pyplot as plt

corte = c.loc[c["semana"] <= 5, "orden"].max() + 0.5

fuera_x = c[c["fuera_x"]]
fuera_r = c[c["fuera_r"]]

fig, ax = plt.subplots(2, 1, figsize=(9, 6), sharex=True)

# Carta X-barra
ax[0].plot(c["orden"], c["xbar"], "o-", markersize=3)
ax[0].axhline(xbb, color="black")
ax[0].axhline(LCSx, color="red", linestyle="--")
ax[0].axhline(LCIx, color="red", linestyle="--")
ax[0].axvline(corte, color="gray")
ax[0].scatter(fuera_x["orden"], fuera_x["xbar"], color="red", zorder=5)
ax[0].text(0.25, 0.92, "Fase I", transform=ax[0].transAxes, ha="center")
ax[0].text(0.75, 0.92, "Fase II", transform=ax[0].transAxes, ha="center")
ax[0].set(ylabel="Media (mL)", title="Carta X-barra, " + linea + " " + str(formato) + " L")

# Carta R
ax[1].plot(c["orden"], c["R"], "o-", markersize=3)
ax[1].axhline(Rb, color="black")
ax[1].axhline(LCSr, color="red", linestyle="--")
ax[1].axvline(corte, color="gray")
ax[1].scatter(fuera_r["orden"], fuera_r["R"], color="red", zorder=5)
ax[1].set(xlabel="Subgrupo (orden en el tiempo)", ylabel="Rango (mL)",
          title="Carta R, " + linea + " " + str(formato) + " L")

plt.tight_layout()
plt.show()

linea = "L2"
formato = 1.5

c = sg[(sg["linea"] == linea) & (sg["formato"] == formato)]
c = c.sort_values("fecha_hora").reset_index(drop=True)
c["orden"] = c.index + 1

f1 = c[c["semana"] <= 5]      # Fase I: se calculan los límites
f2 = c[c["semana"] > 5]       # Fase II: se juzga contra los límites congelados

len(f1), len(f2)

xbb = f1["xbar"].mean()
Rb = f1["R"].mean()

LCSx = xbb + A2 * Rb
LCIx = xbb - A2 * Rb
LCSr = D4 * Rb

fuera = f1[(f1["R"] > LCSr) | (f1["xbar"] > LCSx) | (f1["xbar"] < LCIx)]
print("Ronda 1:", len(fuera), "puntos fuera")

# se sacan los puntos fuera para recalcular
f1 = f1[(f1["R"] <= LCSr) & (f1["xbar"] <= LCSx) & (f1["xbar"] >= LCIx)]

fuera[["orden", "turno", "sku", "xbar", "R"]]

xbb = f1["xbar"].mean()
Rb = f1["R"].mean()

LCSx = xbb + A2 * Rb
LCIx = xbb - A2 * Rb
LCSr = D4 * Rb

fuera = f1[(f1["R"] > LCSr) | (f1["xbar"] > LCSx) | (f1["xbar"] < LCIx)]
print("Ronda 2:", len(fuera), "puntos fuera")

f1 = f1[(f1["R"] <= LCSr) & (f1["xbar"] <= LCSx) & (f1["xbar"] >= LCIx)]

fuera[["orden", "turno", "sku", "xbar", "R"]]

xbb = f1["xbar"].mean()
Rb = f1["R"].mean()

LCSx = xbb + A2 * Rb
LCIx = xbb - A2 * Rb
LCSr = D4 * Rb

fuera = f1[(f1["R"] > LCSr) | (f1["xbar"] > LCSx) | (f1["xbar"] < LCIx)]
print("Ronda 3:", len(fuera), "puntos fuera")

f1 = f1[(f1["R"] <= LCSr) & (f1["xbar"] <= LCSx) & (f1["xbar"] >= LCIx)]

fuera[["orden", "turno", "sku", "xbar", "R"]]

xbb = f1["xbar"].mean()
Rb = f1["R"].mean()

LCSx = xbb + A2 * Rb
LCIx = xbb - A2 * Rb
LCSr = D4 * Rb
LCIr = D3 * Rb
sigma = Rb / d2

print("x doble barra:", round(xbb, 2), " R barra:", round(Rb, 2), " sigma:", round(sigma, 3))
print("LCS X:", round(LCSx, 2), " LCI X:", round(LCIx, 2), " LCS R:", round(LCSr, 2))

# puntos de Fase I que quedaron fuera del cálculo
fase1_completa = c[c["semana"] <= 5]
excluidos = fase1_completa[fase1_completa["orden"].isin(f1["orden"]) == False]

excluidos[["orden", "fecha_hora", "turno", "sku", "xbar", "R"]]

c["fuera_x"] = (c["xbar"] > LCSx) | (c["xbar"] < LCIx)
c["fuera_r"] = (c["R"] > LCSr) | (c["R"] < LCIr)

f2 = c[c["semana"] > 5]

print("Fase II, puntos fuera en X:", f2["fuera_x"].sum(), "de", len(f2))
print("Fase II, puntos fuera en R:", f2["fuera_r"].sum(), "de", len(f2))

resumen.append([linea, formato, len(fase1_completa), len(excluidos),
                xbb, Rb, LCSx, LCIx, LCSr, sigma,
                f2["fuera_x"].sum(), f2["fuera_r"].sum()])

# BLOQUES 3.2 a 3.6 (capacidad, desempeño, límite legal y cifras de esta combinación)
LIE = esp.loc[formato, "LIE_mL"]
LSE = esp.loc[formato, "LSE_mL"]

d = llen[llen["subgrupo"].isin(f1["subgrupo"])]      # mediciones de la Fase I estable

g = d.groupby("subgrupo")["volumen_mL"]
mu, Rb = g.mean().mean(), (g.max() - g.min()).mean()
sigma_c = Rb / d2                                    # corto plazo -> Cp, Cpk
sigma_l = d["volumen_mL"].std(ddof=1)                # largo plazo -> Pp, Ppk

Cp = (LSE - LIE) / (6 * sigma_c)
Cpk = min(LSE - mu, mu - LIE) / (3 * sigma_c)
Pp = (LSE - LIE) / (6 * sigma_l)
Ppk = min(LSE - mu, mu - LIE) / (3 * sigma_l)

print(linea, formato, "Global")
print("Cp =", round(Cp, 3), " Cpk =", round(Cpk, 3), " Pp =", round(Pp, 3), " Ppk =", round(Ppk, 3))

capacidades.append([linea, formato, "Global", len(f1), mu, sigma_c, sigma_l, Cp, Cpk, Pp, Ppk])

for turno in sorted(d["turno"].unique()):
    d_t = d[d["turno"] == turno]

    g = d_t.groupby("subgrupo")["volumen_mL"]
    mu, Rb = g.mean().mean(), (g.max() - g.min()).mean()
    sigma_c = Rb / d2
    sigma_l = d_t["volumen_mL"].std(ddof=1)

    Cp = (LSE - LIE) / (6 * sigma_c)
    Cpk = min(LSE - mu, mu - LIE) / (3 * sigma_c)
    Pp = (LSE - LIE) / (6 * sigma_l)
    Ppk = min(LSE - mu, mu - LIE) / (3 * sigma_l)

    capacidades.append([linea, formato, turno, g.ngroups, mu, sigma_c, sigma_l, Cp, Cpk, Pp, Ppk])

d_f2 = llen[llen["subgrupo"].isin(f2["subgrupo"])]

mu = d_f2["volumen_mL"].mean()
sigma_l = d_f2["volumen_mL"].std(ddof=1)

Pp = (LSE - LIE) / (6 * sigma_l)
Ppk = min(LSE - mu, mu - LIE) / (3 * sigma_l)

desempeno.append([linea, formato, len(f2), mu, sigma_l, Pp, Ppk])

nominal = esp.loc[formato, "nominal_mL"]
T = esp.loc[formato, "tolerancia_oiml_mL"]

x = llen[(llen["linea"] == linea) & (llen["formato_L"] == formato)]["volumen_mL"]   # todas las mediciones

media = x.mean()
error_T1 = ((x < nominal - T) & (x >= nominal - 2 * T)).sum()    # deficiencia mayor que T y hasta 2T
error_T2 = (x < nominal - 2 * T).sum()                           # deficiencia mayor que 2T
pct_T1 = 100 * error_T1 / len(x)

# R87: (1) media >= nominal, (2) a lo más 2,5 % con error T1, (3) ninguno con error T2
cumple = (media >= nominal) and (pct_T1 <= 2.5) and (error_T2 == 0)

legales.append([linea, formato, nominal, T, len(x), media, x.min(), error_T1, pct_T1, error_T2, cumple])

centro = (LSE + LIE) / 2
k_desc = abs(xbb - centro) / ((LSE - LIE) / 2)       # descentrado (lámina 22): 0 = centrado
sigma_meta = (LSE - LIE) / (6 * 1.33)                # sigma necesaria para llegar a Cp = 1,33
reduccion = 100 * (1 - sigma_meta / sigma)           # % que habría que reducir sigma

sobre_f1 = (d["volumen_mL"] > LSE).sum()             # botellas sobre el LSE, Fase I estable
sobre_f2 = (d_f2["volumen_mL"] > LSE).sum()          # botellas sobre el LSE, Fase II
bajo_lie = (x < LIE).sum()                           # botellas bajo el LIE, todas las mediciones

cifras.append([linea, formato, k_desc, sigma, sigma_meta, reduccion,
               len(d), sobre_f1, len(d_f2), sobre_f2, bajo_lie])

corte = c.loc[c["semana"] <= 5, "orden"].max() + 0.5

fuera_x = c[c["fuera_x"]]
fuera_r = c[c["fuera_r"]]

fig, ax = plt.subplots(2, 1, figsize=(9, 6), sharex=True)

# Carta X-barra
ax[0].plot(c["orden"], c["xbar"], "o-", markersize=3)
ax[0].axhline(xbb, color="black")
ax[0].axhline(LCSx, color="red", linestyle="--")
ax[0].axhline(LCIx, color="red", linestyle="--")
ax[0].axvline(corte, color="gray")
ax[0].scatter(fuera_x["orden"], fuera_x["xbar"], color="red", zorder=5)
ax[0].text(0.25, 0.92, "Fase I", transform=ax[0].transAxes, ha="center")
ax[0].text(0.75, 0.92, "Fase II", transform=ax[0].transAxes, ha="center")
ax[0].set(ylabel="Media (mL)", title="Carta X-barra, " + linea + " " + str(formato) + " L")

# Carta R
ax[1].plot(c["orden"], c["R"], "o-", markersize=3)
ax[1].axhline(Rb, color="black")
ax[1].axhline(LCSr, color="red", linestyle="--")
ax[1].axvline(corte, color="gray")
ax[1].scatter(fuera_r["orden"], fuera_r["R"], color="red", zorder=5)
ax[1].set(xlabel="Subgrupo (orden en el tiempo)", ylabel="Rango (mL)",
          title="Carta R, " + linea + " " + str(formato) + " L")

plt.tight_layout()
plt.show()

linea = "L2"
formato = 3.0

c = sg[(sg["linea"] == linea) & (sg["formato"] == formato)]
c = c.sort_values("fecha_hora").reset_index(drop=True)
c["orden"] = c.index + 1

f1 = c[c["semana"] <= 5]      # Fase I: se calculan los límites
f2 = c[c["semana"] > 5]       # Fase II: se juzga contra los límites congelados

len(f1), len(f2)

xbb = f1["xbar"].mean()
Rb = f1["R"].mean()

LCSx = xbb + A2 * Rb
LCIx = xbb - A2 * Rb
LCSr = D4 * Rb

fuera = f1[(f1["R"] > LCSr) | (f1["xbar"] > LCSx) | (f1["xbar"] < LCIx)]
print("Ronda 1:", len(fuera), "puntos fuera")

# se sacan los puntos fuera para recalcular
f1 = f1[(f1["R"] <= LCSr) & (f1["xbar"] <= LCSx) & (f1["xbar"] >= LCIx)]

fuera[["orden", "turno", "sku", "xbar", "R"]]

xbb = f1["xbar"].mean()
Rb = f1["R"].mean()

LCSx = xbb + A2 * Rb
LCIx = xbb - A2 * Rb
LCSr = D4 * Rb

fuera = f1[(f1["R"] > LCSr) | (f1["xbar"] > LCSx) | (f1["xbar"] < LCIx)]
print("Ronda 2:", len(fuera), "puntos fuera")

f1 = f1[(f1["R"] <= LCSr) & (f1["xbar"] <= LCSx) & (f1["xbar"] >= LCIx)]

fuera[["orden", "turno", "sku", "xbar", "R"]]

xbb = f1["xbar"].mean()
Rb = f1["R"].mean()

LCSx = xbb + A2 * Rb
LCIx = xbb - A2 * Rb
LCSr = D4 * Rb

fuera = f1[(f1["R"] > LCSr) | (f1["xbar"] > LCSx) | (f1["xbar"] < LCIx)]
print("Ronda 3:", len(fuera), "puntos fuera")

f1 = f1[(f1["R"] <= LCSr) & (f1["xbar"] <= LCSx) & (f1["xbar"] >= LCIx)]

fuera[["orden", "turno", "sku", "xbar", "R"]]

xbb = f1["xbar"].mean()
Rb = f1["R"].mean()

LCSx = xbb + A2 * Rb
LCIx = xbb - A2 * Rb
LCSr = D4 * Rb
LCIr = D3 * Rb
sigma = Rb / d2

print("x doble barra:", round(xbb, 2), " R barra:", round(Rb, 2), " sigma:", round(sigma, 3))
print("LCS X:", round(LCSx, 2), " LCI X:", round(LCIx, 2), " LCS R:", round(LCSr, 2))

# puntos de Fase I que quedaron fuera del cálculo
fase1_completa = c[c["semana"] <= 5]
excluidos = fase1_completa[fase1_completa["orden"].isin(f1["orden"]) == False]

excluidos[["orden", "fecha_hora", "turno", "sku", "xbar", "R"]]

c["fuera_x"] = (c["xbar"] > LCSx) | (c["xbar"] < LCIx)
c["fuera_r"] = (c["R"] > LCSr) | (c["R"] < LCIr)

f2 = c[c["semana"] > 5]

print("Fase II, puntos fuera en X:", f2["fuera_x"].sum(), "de", len(f2))
print("Fase II, puntos fuera en R:", f2["fuera_r"].sum(), "de", len(f2))

resumen.append([linea, formato, len(fase1_completa), len(excluidos),
                xbb, Rb, LCSx, LCIx, LCSr, sigma,
                f2["fuera_x"].sum(), f2["fuera_r"].sum()])

# BLOQUES 3.2 a 3.6 (capacidad, desempeño, límite legal y cifras de esta combinación)
LIE = esp.loc[formato, "LIE_mL"]
LSE = esp.loc[formato, "LSE_mL"]

d = llen[llen["subgrupo"].isin(f1["subgrupo"])]      # mediciones de la Fase I estable

g = d.groupby("subgrupo")["volumen_mL"]
mu, Rb = g.mean().mean(), (g.max() - g.min()).mean()
sigma_c = Rb / d2                                    # corto plazo -> Cp, Cpk
sigma_l = d["volumen_mL"].std(ddof=1)                # largo plazo -> Pp, Ppk

Cp = (LSE - LIE) / (6 * sigma_c)
Cpk = min(LSE - mu, mu - LIE) / (3 * sigma_c)
Pp = (LSE - LIE) / (6 * sigma_l)
Ppk = min(LSE - mu, mu - LIE) / (3 * sigma_l)

print(linea, formato, "Global")
print("Cp =", round(Cp, 3), " Cpk =", round(Cpk, 3), " Pp =", round(Pp, 3), " Ppk =", round(Ppk, 3))

capacidades.append([linea, formato, "Global", len(f1), mu, sigma_c, sigma_l, Cp, Cpk, Pp, Ppk])

for turno in sorted(d["turno"].unique()):
    d_t = d[d["turno"] == turno]

    g = d_t.groupby("subgrupo")["volumen_mL"]
    mu, Rb = g.mean().mean(), (g.max() - g.min()).mean()
    sigma_c = Rb / d2
    sigma_l = d_t["volumen_mL"].std(ddof=1)

    Cp = (LSE - LIE) / (6 * sigma_c)
    Cpk = min(LSE - mu, mu - LIE) / (3 * sigma_c)
    Pp = (LSE - LIE) / (6 * sigma_l)
    Ppk = min(LSE - mu, mu - LIE) / (3 * sigma_l)

    capacidades.append([linea, formato, turno, g.ngroups, mu, sigma_c, sigma_l, Cp, Cpk, Pp, Ppk])

d_f2 = llen[llen["subgrupo"].isin(f2["subgrupo"])]

mu = d_f2["volumen_mL"].mean()
sigma_l = d_f2["volumen_mL"].std(ddof=1)

Pp = (LSE - LIE) / (6 * sigma_l)
Ppk = min(LSE - mu, mu - LIE) / (3 * sigma_l)

desempeno.append([linea, formato, len(f2), mu, sigma_l, Pp, Ppk])

nominal = esp.loc[formato, "nominal_mL"]
T = esp.loc[formato, "tolerancia_oiml_mL"]

x = llen[(llen["linea"] == linea) & (llen["formato_L"] == formato)]["volumen_mL"]   # todas las mediciones

media = x.mean()
error_T1 = ((x < nominal - T) & (x >= nominal - 2 * T)).sum()    # deficiencia mayor que T y hasta 2T
error_T2 = (x < nominal - 2 * T).sum()                           # deficiencia mayor que 2T
pct_T1 = 100 * error_T1 / len(x)

# R87: (1) media >= nominal, (2) a lo más 2,5 % con error T1, (3) ninguno con error T2
cumple = (media >= nominal) and (pct_T1 <= 2.5) and (error_T2 == 0)

legales.append([linea, formato, nominal, T, len(x), media, x.min(), error_T1, pct_T1, error_T2, cumple])

centro = (LSE + LIE) / 2
k_desc = abs(xbb - centro) / ((LSE - LIE) / 2)       # descentrado (lámina 22): 0 = centrado
sigma_meta = (LSE - LIE) / (6 * 1.33)                # sigma necesaria para llegar a Cp = 1,33
reduccion = 100 * (1 - sigma_meta / sigma)           # % que habría que reducir sigma

sobre_f1 = (d["volumen_mL"] > LSE).sum()             # botellas sobre el LSE, Fase I estable
sobre_f2 = (d_f2["volumen_mL"] > LSE).sum()          # botellas sobre el LSE, Fase II
bajo_lie = (x < LIE).sum()                           # botellas bajo el LIE, todas las mediciones

cifras.append([linea, formato, k_desc, sigma, sigma_meta, reduccion,
               len(d), sobre_f1, len(d_f2), sobre_f2, bajo_lie])

corte = c.loc[c["semana"] <= 5, "orden"].max() + 0.5

fuera_x = c[c["fuera_x"]]
fuera_r = c[c["fuera_r"]]

fig, ax = plt.subplots(2, 1, figsize=(9, 6), sharex=True)

# Carta X-barra
ax[0].plot(c["orden"], c["xbar"], "o-", markersize=3)
ax[0].axhline(xbb, color="black")
ax[0].axhline(LCSx, color="red", linestyle="--")
ax[0].axhline(LCIx, color="red", linestyle="--")
ax[0].axvline(corte, color="gray")
ax[0].scatter(fuera_x["orden"], fuera_x["xbar"], color="red", zorder=5)
ax[0].text(0.25, 0.92, "Fase I", transform=ax[0].transAxes, ha="center")
ax[0].text(0.75, 0.92, "Fase II", transform=ax[0].transAxes, ha="center")
ax[0].set(ylabel="Media (mL)", title="Carta X-barra, " + linea + " " + str(formato) + " L")

# Carta R
ax[1].plot(c["orden"], c["R"], "o-", markersize=3)
ax[1].axhline(Rb, color="black")
ax[1].axhline(LCSr, color="red", linestyle="--")
ax[1].axvline(corte, color="gray")
ax[1].scatter(fuera_r["orden"], fuera_r["R"], color="red", zorder=5)
ax[1].set(xlabel="Subgrupo (orden en el tiempo)", ylabel="Rango (mL)",
          title="Carta R, " + linea + " " + str(formato) + " L")

plt.tight_layout()
plt.show()

columnas = ["linea", "formato", "n_fase1", "excluidos", "x_doble_barra", "R_barra",
            "LCS_x", "LCI_x", "LCS_R", "sigma", "fase2_fuera_x", "fase2_fuera_R"]

tabla = pd.DataFrame(resumen, columns=columnas).round(2)
tabla

# Tabla de límites de cada carta, SIN redondear (sale de la lista `resumen` que se fue llenando arriba)
cols_res = ["linea", "formato", "n_fase1", "excluidos", "xbb", "Rb", "LCSx", "LCIx", "LCSr", "sigma",
            "f2_fuera_x", "f2_fuera_r"]
res = pd.DataFrame(resumen, columns=cols_res)

# las 3 cartas: (línea, formato)
cartas = [(r.linea, r.formato) for r in res.itertuples()]

# especificaciones indexadas por formato (para buscar LIE, LSE, nominal, etc.)
spec_i = spec.set_index("formato_L")

from scipy import stats
res.round(3)



"""## 1. Reglas de decisión y causas asignables

Un punto fuera de los límites no es la única señal de que algo cambió. Las **reglas de Western Electric** buscan patrones poco probables si solo actúan causas comunes. Se aplican a la carta X̄ usando la media estandarizada

$$z=\frac{\bar{x}_i-\bar{\bar{x}}}{\sigma/\sqrt{n}}$$

con los límites congelados de Fase I (así no se contamina el criterio con datos de Fase II):

| Regla | Señal |
|---|---|
| R1 | 1 punto a más de 3 sigma de la línea central |
| R2 | 2 de 3 puntos seguidos a más de 2 sigma, del mismo lado |
| R3 | 4 de 5 puntos seguidos a más de 1 sigma, del mismo lado |
| R4 | 8 puntos seguidos del mismo lado de la línea central |
| R alta | un rango por sobre el límite superior de la carta R |
"""

# Reglas de Western Electric sobre la carta X-barra.
# z = cuántos "sigma de la media" está cada subgrupo de la línea central: z = (xbar - xbb) / (sigma / raiz(n))
def reglas_we(z):
    r1 = z.abs() > 3                                                              # 1 punto a más de 3 sigma
    r2 = ((z > 2).astype(int).rolling(3).sum() >= 2) | ((z < -2).astype(int).rolling(3).sum() >= 2)   # 2 de 3 pasan 2 sigma (mismo lado)
    r3 = ((z > 1).astype(int).rolling(5).sum() >= 4) | ((z < -1).astype(int).rolling(5).sum() >= 4)   # 4 de 5 pasan 1 sigma (mismo lado)
    r4 = ((z > 0).astype(int).rolling(8).sum() == 8) | ((z < 0).astype(int).rolling(8).sum() == 8)    # 8 seguidos del mismo lado
    return r1, r2, r3, r4

senales = {}
filas = []

for linea, formato in cartas:
    # 1. Obtener los límites de Fase I para esta combinación de línea y formato
    es_esta_carta = (res["linea"] == linea) & (res["formato"] == formato)
    lim = res[es_esta_carta].iloc[0]

    # 2. Filtrar, ordenar y numerar los subgrupos correspondientes
    c = sg[(sg["linea"] == linea) & (sg["formato"] == formato)].copy()
    c = c.sort_values("fecha_hora").reset_index(drop=True)
    c["orden"] = c.index + 1
    c["fase"] = np.where(c["semana"] <= 5, "I", "II")

    # 3. Calcular el puntaje Z (distancia a la media en unidades de error estándar)
    error_estandar = lim["sigma"] / np.sqrt(5)
    c["z"] = (c["xbar"] - lim["xbb"]) / error_estandar

    # 4. Evaluar las reglas de Western Electric y alarmas en la carta R
    c["R1"], c["R2"], c["R3"], c["R4"] = reglas_we(c["z"])
    c["R_alta"] = c["R"] > lim["LCSr"]

    # 5. Si se activa cualquiera de las reglas, se considera que hay señal/alarma
    columnas_alarmas = ["R1", "R2", "R3", "R4", "R_alta"]
    c["senal"] = c[columnas_alarmas].any(axis=1)

    # Guardamos los datos de esta combinación
    senales[(linea, formato)] = c

    # 6. Agrupar resultados por Fase (I y II) para el resumen
    for fase in ["I", "II"]:
        datos_fase = c[c["fase"] == fase]

        nombre_carta = f"{linea} {formato} L"
        filas.append([
            nombre_carta,
            fase,
            len(datos_fase),
            datos_fase["R1"].sum(),
            datos_fase["R2"].sum(),
            datos_fase["R3"].sum(),
            datos_fase["R4"].sum(),
            datos_fase["R_alta"].sum(),
            datos_fase["senal"].sum()
        ])

# 7. Construir y mostrar el DataFrame final
columnas_resumen = [
    "carta", "fase", "subgrupos",
    "R1 (>3 sigma)", "R2 (2 de 3)", "R3 (4 de 5)", "R4 (8 seguidos)",
    "R alta (carta R)", "subgrupos con señal"
]
tabla_we = pd.DataFrame(filas, columns=columnas_resumen)
tabla_we

# 1. Inicializar listas para recopilar las métricas de señales por diferentes dimensiones
resumen_turno = []
resumen_sku = []
resumen_semana = []

# 2. Iterar sobre cada carta de control y sus datos de señales procesados
for (linea, formato), datos_carta in senales.items():
    nombre_carta = f"{linea} {formato} L"

    # Analizar señales agrupadas por Turno
    for turno, datos_grupo in datos_carta.groupby("turno"):
        total_subgrupos = len(datos_grupo)
        con_senal = datos_grupo["senal"].sum()
        pct_senal = 100 * datos_grupo["senal"].mean()

        resumen_turno.append([
            nombre_carta, turno, total_subgrupos, con_senal,
            pct_senal, datos_grupo["xbar"].mean(), datos_grupo["R"].mean()
        ])

    # Analizar señales agrupadas por SKU
    for sku, datos_grupo in datos_carta.groupby("sku"):
        total_subgrupos = len(datos_grupo)
        con_senal = datos_grupo["senal"].sum()
        pct_senal = 100 * datos_grupo["senal"].mean()

        resumen_sku.append([
            nombre_carta, sku, total_subgrupos, con_senal,
            pct_senal, datos_grupo["xbar"].mean(), datos_grupo["R"].mean()
        ])

    # Analizar señales agrupadas por Semana
    for semana, datos_grupo in datos_carta.groupby("semana"):
        resumen_semana.append([
            nombre_carta, semana, datos_grupo["senal"].sum(),
            datos_grupo["xbar"].mean(), datos_grupo["R"].mean()
        ])

# 3. Crear los DataFrames resultantes con nombres de columnas claros
columnas_analisis = ["carta", "grupo", "subgrupos", "con_señal", "%_con_señal", "xbar_medio", "R_medio"]
por_turno = pd.DataFrame(resumen_turno, columns=columnas_analisis)
por_sku = pd.DataFrame(resumen_sku, columns=columnas_analisis)
por_sem = pd.DataFrame(resumen_semana, columns=["carta", "semana", "con_señal", "xbar_medio", "R_medio"])

# 4. Mostrar los reportes finales formateados
print("=== SEÑALES POR TURNO ===")
display(por_turno.round(2))

print("\n=== SEÑALES POR SKU ===")
display(por_sku.round(2))

print("\n=== MEDIA DE X-BARRA POR SEMANA (Línea central Fase I: Semanas 1-5) ===")
media_semanal_pivot = por_sem.pivot(index="carta", columns="semana", values="xbar_medio")
display(media_semanal_pivot.round(1))

print("\n=== SUBGRUPOS CON SEÑAL POR SEMANA ===")
senales_semanal_pivot = por_sem.pivot(index="carta", columns="semana", values="con_señal")
display(senales_semanal_pivot)

import matplotlib.pyplot as plt
import seaborn as sns

# Configurar estilo visual
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 6))

# Crear el gráfico de barras agrupado por semana y carta
ax = sns.barplot(
    data=por_sem,
    x="semana",
    y="con_señal",
    hue="carta",
    palette="viridis"
)

# Añadir títulos y etiquetas claras
plt.title("Cantidad de Subgrupos con Señal de Control por Semana", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Semana de Registro", fontsize=12)
plt.ylabel("Subgrupos con Señal", fontsize=12)
plt.legend(title="Carta de Control", title_fontsize="11", loc="upper left")

# Añadir el valor numérico arriba de cada barra para facilitar la lectura
for container in ax.containers:
    ax.bar_label(container, fmt="%d", padding=3, fontsize=9)

plt.tight_layout()
plt.show()

import matplotlib.pyplot as plt
import seaborn as sns

# Configurar el estilo visual
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 6))

# Crear el gráfico de barras agrupado por turno y carta de control
ax = sns.barplot(
    data=por_turno,
    x="grupo",
    y="con_señal",
    hue="carta",
    palette="magma"
)

# Personalizar el gráfico con etiquetas y títulos claros
plt.title("Subgrupos con Señal de Control por Turno de Trabajo", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Turno", fontsize=12)
plt.ylabel("Cantidad de Subgrupos con Señal", fontsize=12)
plt.legend(title="Carta de Control", title_fontsize="11", loc="upper right")

# Agregar los valores numéricos sobre las barras para facilitar la lectura rápida
for container in ax.containers:
    ax.bar_label(container, fmt="%d", padding=3, fontsize=10)

plt.tight_layout()
plt.show()

# 1. Consolidar todos los datos de señales de las diferentes cartas en un solo DataFrame
todos_los_subgrupos = pd.concat(senales.values(), ignore_index=True)

# 2. Agrupar por SKU para calcular las métricas clave de señales
resumen_critico_sku = todos_los_subgrupos.groupby("sku").agg(
    total_subgrupos=("senal", "count"),
    subgrupos_con_senal=("senal", "sum"),
    porcentaje_con_senal=("senal", lambda x: 100 * x.mean())
).reset_index()

# 3. Ordenar de mayor a menor porcentaje para destacar los SKUs críticos
resumen_critico_sku = resumen_critico_sku.sort_values(by="porcentaje_con_senal", ascending=False).reset_index(drop=True)

# 4. Mostrar el reporte final
print("=== ANÁLISIS DE SKUs CRÍTICOS (ORDENADO POR % DE SEÑALES) ===")
display(resumen_critico_sku.round(2))

import matplotlib.pyplot as plt
import seaborn as sns

# Configurar el estilo visual
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 6))

# Crear el gráfico de barras para subgrupos con señal por SKU
ax = sns.barplot(
    data=resumen_critico_sku,
    x="sku",
    y="subgrupos_con_senal",
    palette="Reds_r"
)

# Personalizar el gráfico
plt.title("Cantidad de Subgrupos con Señal de Control por SKU (Productos Críticos)", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("SKU (Producto)", fontsize=12)
plt.ylabel("Subgrupos con Señal", fontsize=12)

# Añadir los valores numéricos sobre las barras para facilitar la lectura
for container in ax.containers:
    ax.bar_label(container, fmt="%d", padding=3, fontsize=10)

plt.tight_layout()
plt.show()

# 1. Agrupar por Turno y SKU para identificar combinaciones críticas
resumen_turno_sku = todos_los_subgrupos.groupby(["turno", "sku"]).agg(
    total_subgrupos=("senal", "count"),
    subgrupos_con_senal=("senal", "sum"),
    porcentaje_con_senal=("senal", lambda x: 100 * x.mean())
).reset_index()

# 2. Ordenar por criticidad (porcentaje de señales descendente)
resumen_turno_sku = resumen_turno_sku.sort_values(by="porcentaje_con_senal", ascending=False).reset_index(drop=True)

# 3. Mostrar la tabla resultante
print("=== ANÁLISIS CRÍTICO: SKU POR TURNO (ORDENADO POR % DE SEÑALES) ===")
display(resumen_turno_sku.round(2))

# En L2 los SKU no se reparten igual entre turnos; hay que ver si el "efecto SKU" es solo el efecto del turno.
# Se compara la media dentro de cada turno (solo Fase I, antes de cualquier cambio de nivel)
c = senales[("L2", 1.5)]
c = c[c["fase"] == "I"]
print("L2 1,5 L, Fase I: media de X-barra por turno y SKU")
display(c.pivot_table(index="sku", columns="turno", values="xbar", aggfunc="mean").round(1))
print("Cantidad de subgrupos")
display(c.pivot_table(index="sku", columns="turno", values="xbar", aggfunc="count"))

import matplotlib.pyplot as plt

# 1. Gráfico independiente para L1 0.5 L
c = senales[("L1", 0.5)]
fig, ax = plt.subplots(figsize=(10, 3.5))

ax.plot(c["orden"], c["z"], "o-", color="steelblue", ms=3)
ax.axhline(0, color="black", lw=1)

# Líneas de zonas de control (+/- 1, 2, 3 sigma)
for z_lim, est_linea, color_linea in [(3, "--", "red"), (-3, "--", "red"), (2, ":", "gray"), (-2, ":", "gray"), (1, ":", "gray"), (-1, ":", "gray")]:
    ax.axhline(z_lim, linestyle=est_linea, color=color_linea)

# Destacar puntos con señal activa
c_con_senal = c[c["senal"]]
ax.scatter(c_con_senal["orden"], c_con_senal["z"], color="red", s=14, zorder=5)

# Línea vertical divisoria entre Fase I y Fase II
ax.axvline(c["orden"][c["fase"] == "I"].max() + 0.5, color="gray")
ax.set_ylabel("z de X-barra")
ax.set_title("Reglas Western Electric: L1 0.5 L (Rojo = con señal)", fontsize=11, fontweight="bold")
ax.set_xlabel("Subgrupo (orden en el tiempo)")

plt.tight_layout()
plt.show()

# 2. Gráfico independiente para L2 1.5 L
c = senales[("L2", 1.5)]
fig, ax = plt.subplots(figsize=(10, 3.5))

ax.plot(c["orden"], c["z"], "o-", color="steelblue", ms=3)
ax.axhline(0, color="black", lw=1)

# Líneas de zonas de control (+/- 1, 2, 3 sigma)
for z_lim, est_linea, color_linea in [(3, "--", "red"), (-3, "--", "red"), (2, ":", "gray"), (-2, ":", "gray"), (1, ":", "gray"), (-1, ":", "gray")]:
    ax.axhline(z_lim, linestyle=est_linea, color=color_linea)

# Destacar puntos con señal activa
c_con_senal = c[c["senal"]]
ax.scatter(c_con_senal["orden"], c_con_senal["z"], color="red", s=14, zorder=5)

# Línea vertical divisoria entre Fase I y Fase II
ax.axvline(c["orden"][c["fase"] == "I"].max() + 0.5, color="gray")
ax.set_ylabel("z de X-barra")
ax.set_title("Reglas Western Electric: L2 1.5 L (Rojo = con señal)", fontsize=11, fontweight="bold")
ax.set_xlabel("Subgrupo (orden en el tiempo)")

plt.tight_layout()
plt.show()

# 3. Gráfico independiente para L2 3.0 L
c = senales[("L2", 3.0)]
fig, ax = plt.subplots(figsize=(10, 3.5))

ax.plot(c["orden"], c["z"], "o-", color="steelblue", ms=3)
ax.axhline(0, color="black", lw=1)

# Líneas de zonas de control (+/- 1, 2, 3 sigma)
for z_lim, est_linea, color_linea in [(3, "--", "red"), (-3, "--", "red"), (2, ":", "gray"), (-2, ":", "gray"), (1, ":", "gray"), (-1, ":", "gray")]:
    ax.axhline(z_lim, linestyle=est_linea, color=color_linea)

# Destacar puntos con señal activa
c_con_senal = c[c["senal"]]
ax.scatter(c_con_senal["orden"], c_con_senal["z"], color="red", s=14, zorder=5)

# Línea vertical divisoria entre Fase I y Fase II
ax.axvline(c["orden"][c["fase"] == "I"].max() + 0.5, color="gray")
ax.set_ylabel("z de X-barra")
ax.set_title("Reglas Western Electric: L2 3.0 L (Rojo = con señal)", fontsize=11, fontweight="bold")
ax.set_xlabel("Subgrupo (orden en el tiempo)")

plt.tight_layout()
plt.show()

"""### Análisis ANOVA para L1 0.5 L"""

f1_l1 = senales[("L1", 0.5)]
f1_l1 = f1_l1[f1_l1["fase"] == "I"]

g_turno_l1 = [g["xbar"] for _, g in f1_l1.groupby("turno")]
g_sku_l1 = [g["xbar"] for _, g in f1_l1.groupby("sku")]

p_turno_l1 = stats.f_oneway(*g_turno_l1)[1]
p_sku_l1 = stats.f_oneway(*g_sku_l1)[1]

print(f"L1 0.5 L  p-valor turno = {p_turno_l1:.4f} | p-valor SKU = {p_sku_l1:.4f}")

"""### Análisis ANOVA para L2 1.5 L"""

f1_l2_15 = senales[("L2", 1.5)]
f1_l2_15 = f1_l2_15[f1_l2_15["fase"] == "I"]

g_turno_l2_15 = [g["xbar"] for _, g in f1_l2_15.groupby("turno")]
g_sku_l2_15 = [g["xbar"] for _, g in f1_l2_15.groupby("sku")]

p_turno_l2_15 = stats.f_oneway(*g_turno_l2_15)[1]
p_sku_l2_15 = stats.f_oneway(*g_sku_l2_15)[1]

print(f"L2 1.5 L  p-valor turno = {p_turno_l2_15:.4f} | p-valor SKU = {p_sku_l2_15:.4f}")

"""### Análisis ANOVA para L2 3.0 L"""

f1_l2_30 = senales[("L2", 3.0)]
f1_l2_30 = f1_l2_30[f1_l2_30["fase"] == "I"]

g_turno_l2_30 = [g["xbar"] for _, g in f1_l2_30.groupby("turno")]
g_sku_l2_30 = [g["xbar"] for _, g in f1_l2_30.groupby("sku")]

p_turno_l2_30 = stats.f_oneway(*g_turno_l2_30)[1]
p_sku_l2_30 = stats.f_oneway(*g_sku_l2_30)[1]

print(f"L2 3.0 L  p-valor turno = {p_turno_l2_30:.4f} | p-valor SKU = {p_sku_l2_30:.4f}")


columnas_cap = ["linea", "formato", "grupo", "n_subgrupos", "media", "sigma_corto",
                "sigma_largo", "Cp", "Cpk", "Pp", "Ppk"]

tabla_cap = pd.DataFrame(capacidades, columns=columnas_cap)
tabla_cap["Cp/Cpk"] = tabla_cap["Cp"] / tabla_cap["Cpk"]    # 1 = centrado

display(tabla_cap.round(2))

tabla_f2 = pd.DataFrame(desempeno, columns=["linea", "formato", "n_subgrupos", "media",
                                            "sigma_largo", "Pp", "Ppk"])

display(tabla_f2.round(2))

columnas_legal = ["linea", "formato", "nominal", "T", "botellas", "media", "minimo",
                  "error_T1", "pct_T1", "error_T2", "cumple_R87"]

tabla_legal = pd.DataFrame(legales, columns=columnas_legal)
tabla_legal["media_menos_nominal"] = tabla_legal["media"] - tabla_legal["nominal"]
tabla_legal["margen_minimo"] = tabla_legal["minimo"] - (tabla_legal["nominal"] - tabla_legal["T"])

display(tabla_legal.round(2))

columnas_cifras = ["linea", "formato", "k_descentrado", "sigma_actual", "sigma_para_1.33",
                   "reduccion_%", "botellas_F1", "sobre_LSE_F1", "botellas_F2", "sobre_LSE_F2", "bajo_LIE_total"]

tabla_cifras = pd.DataFrame(cifras, columns=columnas_cifras)
tabla_cifras["%_sobre_LSE_F1"] = 100 * tabla_cifras["sobre_LSE_F1"] / tabla_cifras["botellas_F1"]
tabla_cifras["%_sobre_LSE_F2"] = 100 * tabla_cifras["sobre_LSE_F2"] / tabla_cifras["botellas_F2"]

display(tabla_cifras.round(2))

etiquetas = tabla_cap["linea"] + " " + tabla_cap["formato"].astype(str) + " " + tabla_cap["grupo"].str[0]

fig, ax = plt.subplots(figsize=(9, 4.5))

ax.bar(etiquetas, tabla_cap["Cpk"])
ax.axhline(1.33, color="red", linestyle="--")
ax.tick_params(axis="x", rotation=45)
ax.set(ylabel="Cpk", title="Cpk por línea, formato y turno (G = global)")

plt.tight_layout()
plt.show()

##CARTA DE CONTROL p POR LÍNEA CON LÍMITES VARIABLES

# Copia de trabajo de los datos de atributos
df_atr_proc = atr.copy()
resumen_lineas = []

# Se evalúa cada línea por separado (L1 y L2)
for linea_nom, datos in df_atr_proc.groupby("linea"):
    datos = datos.sort_values(["fecha", "turno"]).reset_index(drop=True)
    datos["orden"] = datos.index + 1

    # Proporción observada por subgrupo
    datos["p"] = datos["n_defectuosas"] / datos["n_inspeccionadas"]

    # Totales y cálculo de p-barra para la línea
    tot_insp = datos["n_inspeccionadas"].sum()
    tot_def = datos["n_defectuosas"].sum()
    p_bar = tot_def / tot_insp

    # Límites variables porque el tamaño de inspección n no es constante
    error = np.sqrt((p_bar * (1 - p_bar)) / datos["n_inspeccionadas"])
    datos["LCS"] = np.minimum(1.0, p_bar + 3 * error)
    datos["LCI"] = np.maximum(0.0, p_bar - 3 * error)

    # Detección de puntos fuera de control
    datos["fuera"] = (datos["p"] > datos["LCS"]) | (datos["p"] < datos["LCI"])
    pts_fuera = datos[datos["fuera"]]

    resumen_lineas.append([
        linea_nom,
        tot_insp,
        tot_def,
        round(p_bar * 100, 2),
        len(pts_fuera)
    ])

    # Gráfico de la carta p con límites escalonados
    fig, ax = plt.subplots(figsize=(10, 4))
    ax.plot(datos["orden"], datos["p"], "o-", markersize=3, color="tab:blue", label="Proporción (p)")
    ax.axhline(p_bar, color="black", linestyle="-", label=f"Línea Central (p-barra = {p_bar:.4f})")
    ax.step(datos["orden"], datos["LCS"], color="red", linestyle="--", where="mid", label="LCS variable")
    ax.step(datos["orden"], datos["LCI"], color="red", linestyle=":", where="mid", label="LCI variable")

    if len(pts_fuera) > 0:
        ax.scatter(pts_fuera["orden"], pts_fuera["p"], color="red", s=40, zorder=5, label="Fuera de control")

    ax.set_title(f"Carta de Control p: Línea {linea_nom} (Límites Variables)", fontsize=11)
    ax.set_xlabel("Muestra de Inspección (Orden cronológico)", fontsize=9)
    ax.set_ylabel("Proporción defectuosa (p)", fontsize=9)
    ax.grid(True, linestyle=":", alpha=0.5)
    ax.legend(loc="upper right")
    plt.tight_layout()
    plt.show()

# Tabla resumen por línea
cols_res_p = ["linea", "inspeccionadas", "defectuosas", "%_defectuosas", "puntos_fuera"]
tabla_resumen_p = pd.DataFrame(resumen_lineas, columns=cols_res_p)
print("=== RESUMEN CARTA p POR LÍNEA ===")
display(tabla_resumen_p)

## TABLA DE FRECUENCIAS DE PARETO (EXCLUYENDO 'Sin defectos')

# Según la bitácora, se excluyen las filas registradas como "Sin defectos"
pareto_data = atr[atr["causa_principal"] != "Sin defectos"].copy()

# Suma de botellas defectuosas por cada causa principal
pareto_df = pareto_data.groupby("causa_principal")["n_defectuosas"].sum().reset_index()
pareto_df = pareto_df.sort_values(by="n_defectuosas", ascending=False).reset_index(drop=True)

# Cálculo de porcentaje relativo y acumulado
pareto_df["porcentaje"] = (pareto_df["n_defectuosas"] / pareto_df["n_defectuosas"].sum()) * 100
pareto_df["acumulado"] = pareto_df["porcentaje"].cumsum()

# Presentación de la tabla
pareto_df.columns = ["causa_principal", "defectuosas_asociadas", "%_relativo", "%_acumulado"]
print("\n=== TABLA DE PARETO DE CAUSAS DE DEFECTO ===")
display(pareto_df.round(2))

##GRAFICO DIAGRAMA DE PARETO

fig, ax1 = plt.subplots(figsize=(9, 4.5))

# Barras de cantidad absoluta de defectos
barras = ax1.bar(pareto_df["causa_principal"], pareto_df["defectuosas_asociadas"],
                 color="steelblue", width=0.55)
ax1.set_ylabel("Botellas defectuosas asociadas", color="steelblue", fontsize=10)
ax1.tick_params(axis="x", rotation=15)
ax1.set_xlabel("Causa principal registrada", fontsize=10)

# Curva acumulada en eje secundario
ax2 = ax1.twinx()
ax2.plot(pareto_df["causa_principal"], pareto_df["%_acumulado"],
         color="crimson", marker="o", linewidth=2)
ax2.set_ylabel("Porcentaje acumulado (%)", color="crimson", fontsize=10)
ax2.set_ylim(0, 105)

# Línea de referencia del 80% (corte de Pareto)
ax2.axhline(80, color="gray", linestyle="--", alpha=0.7)
ax2.text(0.5, 82, "Corte 80%", color="gray", fontsize=9)

# Valores numéricos sobre cada barra
for b in barras:
    h = b.get_height()
    ax1.annotate(f"{int(h)}",
                 xy=(b.get_x() + b.get_width() / 2, h),
                 xytext=(0, 3), textcoords="offset points",
                 ha="center", va="bottom", fontsize=9)

plt.title("Diagrama de Pareto: Causas Principales de Defectos en Envasado", fontsize=11)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# ÍTEM 5: CUANTIFICACIÓN ECONÓMICA DEL SOBRELLENADO
# ============================================================

# --- 1. Datos de apoyo ---
mae = maestro.set_index("sku")                    # costo variable, botellas y litros por caja de cada SKU

# ventas del último año (52 semanas) para llevar el sobrellenado a litros al año
ult = dem[dem["semana"] > 104]
cajas_anio = ult.groupby("sku")["ventas_cajas"].sum()    # los 3 datos faltantes de demanda no se suman

# --- 2. Sobrellenado por SKU, en dos escenarios ---
# "Todas las mediciones": media de las 10 semanas. "Fase II": situación actual (semanas 6 a 10)
escenarios = [("Todas las mediciones", llen),
              ("Fase II (semanas 6-10)", llen[llen["semana_registro"] > 5])]

economia = []

for nombre, d in escenarios:
    for sku in mae.index:
        formato = mae.loc[sku, "formato_L"]
        nominal = esp.loc[formato, "nominal_mL"]

        x = d[d["sku"] == sku]["volumen_mL"]
        sobre = x.mean() - nominal                              # mL regalados por botella

        botellas = cajas_anio[sku] * mae.loc[sku, "botellas_por_caja"]
        litros = botellas * sobre / 1000                        # litros regalados al año

        costo_litro = mae.loc[sku, "costo_variable_caja_clp"] / mae.loc[sku, "litros_por_caja"]
        valor = litros * costo_litro                            # CLP al año a costo variable

        economia.append([nombre, mae.loc[sku, "linea_envasado"], formato, sku,
                         sobre, botellas, litros, costo_litro, valor])

columnas_eco = ["escenario", "linea", "formato", "sku", "sobrellenado_mL",
                "botellas_anio", "litros_anio", "costo_var_por_L", "valor_CLP_anio"]

tabla_eco = pd.DataFrame(economia, columns=columnas_eco)

display(tabla_eco.round(2))

# --- 3. Totales por línea y formato, y total de la planta ---
tot_formato = tabla_eco.groupby(["escenario", "linea", "formato"])[["litros_anio", "valor_CLP_anio"]].sum()

display(tot_formato.round(0))

tot_planta = tabla_eco.groupby("escenario")[["litros_anio", "valor_CLP_anio"]].sum()
tot_planta["valor_millones_CLP"] = tot_planta["valor_CLP_anio"] / 1e6

display(tot_planta.round(1))

# --- 4. Desagregado por turno (misma lógica, sin separar por SKU) ---
# Supuesto: cada turno produce en proporción a sus subgrupos muestreados (se muestrea cada 2 horas)
turnos = []

for linea in ["L1", "L2"]:
    for formato in sorted(llen[llen["linea"] == linea]["formato_L"].unique()):
        nominal = esp.loc[formato, "nominal_mL"]
        d = llen[(llen["linea"] == linea) & (llen["formato_L"] == formato)]

        skus = mae[(mae["linea_envasado"] == linea) & (mae["formato_L"] == formato)].index
        botellas = (cajas_anio[skus] * mae.loc[skus, "botellas_por_caja"]).sum()
        costo_litro = (mae.loc[skus, "costo_variable_caja_clp"].sum() / mae.loc[skus, "litros_por_caja"].sum())

        n_sub = d.groupby("turno")["subgrupo"].nunique()        # subgrupos por turno
        for turno in sorted(d["turno"].unique()):
            x = d[d["turno"] == turno]["volumen_mL"]
            sobre = x.mean() - nominal
            botellas_t = botellas * n_sub[turno] / n_sub.sum()  # botellas del turno
            litros = botellas_t * sobre / 1000
            turnos.append([linea, formato, turno, sobre, botellas_t, litros, litros * costo_litro])

tabla_turno = pd.DataFrame(turnos, columns=["linea", "formato", "turno", "sobrellenado_mL",
                                            "botellas_anio", "litros_anio", "valor_CLP_anio"])

display(tabla_turno.round(2))
```

```python
#quiebres estructurales -------------------------------------------
# Se quita la estacionalidad y se busca el corte que mejor separa la serie en dos niveles constantes
desest = limpia / indice_mes.loc[mes].values
filas = []
for sku in skus:
    y = desest[sku].values
    mejor_sse, mejor_t = np.inf, None
    for t in range(26, 131):                # el corte deja al menos 26 semanas a cada lado
        sse = ((y[:t] - y[:t].mean()) ** 2).sum() + ((y[t:] - y[t:].mean()) ** 2).sum()
        if sse < mejor_sse:
            mejor_sse, mejor_t = sse, t
    antes, despues = y[:mejor_t].mean(), y[mejor_t:].mean()
    mejora = 1 - mejor_sse / ((y - y.mean()) ** 2).sum()      # cuánta variación explica el corte
    filas.append([sku, mejor_t + 1, fechas.iloc[mejor_t].date(), antes, despues, (despues / antes - 1) * 100, mejora * 100])

quiebre = pd.DataFrame(filas, columns=["sku", "semana_corte", "fecha", "nivel_antes", "nivel_despues", "cambio_%", "variacion_explicada_%"])
quiebre["hay_quiebre"] = (quiebre["cambio_%"].abs() > 20) & (quiebre["variacion_explicada_%"] > 25)

for _, f in quiebre[quiebre["hay_quiebre"]].iterrows():
    anotar(f["sku"], int(f["semana_corte"]), "Quiebre estructural (nivel " + str(round(f["cambio_%"])) + " %)",
           "Corte que mejor separa dos niveles en la serie sin estacionalidad",
           "Se conserva el dato; el modelo debe seguir el nivel nuevo (pesar los datos recientes)")

quiebre.round(1)
```

```python
#intermitencia ---------------------------------------------------
# ADI = semanas promedio entre ventas; CV² = variabilidad de las ventas cuando sí hay venta
filas = []
for sku in skus:
    v = limpia[sku]
    con_venta = v[v > 0]
    adi = len(v) / len(con_venta)
    cv2 = (con_venta.std() / con_venta.mean()) ** 2
    if adi > 1.32 and cv2 < 0.49:
        tipo = "intermitente"
    elif adi > 1.32:
        tipo = "irregular (grumosa)"
    elif cv2 > 0.49:
        tipo = "errática"
    else:
        tipo = "suave"
    filas.append([sku, (v == 0).sum(), adi, cv2, tipo])

intermit = pd.DataFrame(filas, columns=["sku", "semanas_en_cero", "ADI", "CV2", "tipo"])

# ¿los ceros son quiebres de stock o simplemente no hubo demanda?
ceros_con_quiebre = ((ventas["FUN-B30"] == 0) & (sin_stock["FUN-B30"] > 0)).sum()
print("FUN-B30: semanas en cero con días sin stock:", ceros_con_quiebre)
anotar("FUN-B30", "varias", "Demanda intermitente (" + str(int((limpia["FUN-B30"] == 0).sum())) + " semanas en cero)",
       "ADI = 1.8 (> 1.32) y CV² bajo (< 0.49)", "Se conservan los ceros; no se tratan como atípicos")

intermit.round(2)

```

```python
#### CELDA 1: librerías y partición temporal ----------------------------------------
import warnings
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from statsmodels.tsa.statespace.sarimax import SARIMAX
warnings.filterwarnings("ignore")              # statsmodels avisa mucho sobre convergencia

H = 13                                         # horizonte pedido (13 semanas)
S = 52                                         # largo de la estación (52 semanas = 1 año)
FIN_TRAIN = 156 - H                            # semana 143

train = demanda_estimada.loc[:FIN_TRAIN]                # semanas 1 a 143: ajuste y selección de modelos
test = demanda_estimada.loc[FIN_TRAIN + 1:]             # semanas 144 a 156: prueba, NO se toca para elegir nada
promo_train = promo.loc[:FIN_TRAIN]
promo_test = promo.loc[FIN_TRAIN + 1:]

print("Entrenamiento: semanas", train.index.min(), "a", train.index.max(), "(", len(train), "semanas )")
print("Prueba:        semanas", test.index.min(), "a", test.index.max(), "(", len(test), "semanas )")
assert train.index.max() < test.index.min()    # no hay traslape entre entrenamiento y prueba


#### CELDA 2: ¿promoción y precio como regresores? (se justifica, no se asume) ---------------
# a) Precio de lista: ¿cambia en el tiempo?
print("Valores distintos del precio de lista por SKU:")
print(precio_li.nunique())

# b) Precio efectivo: ¿se mueve solo cuando hay promoción?
descuento = 1 - precio_ef / precio_li
print("Semanas con descuento SIN promoción:", int(((descuento > 0) & (promo == 0)).sum().sum()))
print("Descuento en semanas con promoción: media", round(descuento[promo == 1].stack().mean() * 100, 1),
      "%  | desviación", round(descuento[promo == 1].stack().std() * 100, 1), "%")

# c) ¿la promoción mueve las ventas? (lift del punto 1)
efecto_promo[["sku", "lift_medio", "lift_min", "lift_max"]].round(2)
# El precio de lista no varía y el precio efectivo solo cambia en promoción con un descuento casi fijo:
# precio y promoción son la MISMA información -> se usa solo `promocion` como regresor.


#### CELDA 3: modelo 1, benchmark ingenuo estacional ---------------------------------------
# Pronóstico = lo que pasó la misma semana del año anterior. Es el benchmark adecuado porque
# las series tienen estacionalidad anual fuerte (punto 1); un ingenuo simple ignoraría el verano.
def pron_naive_est(y, h, promo_fut=None):
    return y.iloc[-S:].iloc[:h].values


#### CELDA 4: modelo 2, ETS (Holt-Winters) -----------------------------------------------------
# Nivel + tendencia amortiguada + estacionalidad aditiva de 52 semanas.
# Aditiva (no multiplicativa) porque FUN-B30 tiene ceros y la multiplicativa no los acepta.
def pron_ets(y, h, promo_fut=None):
    modelo = ExponentialSmoothing(y.values, trend="add", damped_trend=True,
                                  seasonal="add", seasonal_periods=S,
                                  initialization_method="estimated").fit()
    return np.maximum(modelo.forecast(h), 0)   # la demanda no puede ser negativa


#### CELDA 5: modelo 3, SARIMA y SARIMAX ----------------------------------------------------------
# SARIMA(p,0,q)(0,1,0)52: la diferencia estacional (D=1) compara cada semana con la del año anterior,
# y la parte ARMA corrige lo que queda. Con estación de 52 semanas y solo 3 años de datos,
# no se agregan términos estacionales extra (no habría datos para estimarlos).
# p y q se eligen por AIC SOLO con el entrenamiento (nunca mirando la prueba).

def elegir_orden(y, exog=None):
    mejor_aic, mejor = np.inf, None
    for p in [0, 1, 2]:
        for q in [0, 1]:
            for tendencia in ["n", "c"]:
                try:
                    m = SARIMAX(y.values, exog=exog, order=(p, 0, q), seasonal_order=(0, 1, 0, S),
                                trend=tendencia).fit(disp=False)
                    if m.aic < mejor_aic:
                        mejor_aic, mejor = m.aic, (p, q, tendencia)
                except Exception:
                    pass
    return mejor

def pron_sarima(y, h, orden, promo_hist=None, promo_fut=None):
    p, q, tendencia = orden
    modelo = SARIMAX(y.values, exog=promo_hist, order=(p, 0, q), seasonal_order=(0, 1, 0, S),
                     trend=tendencia).fit(disp=False)
    return np.maximum(modelo.forecast(h, exog=promo_fut), 0), modelo


#### CELDA 6: elegir órdenes y revisar si la promoción es significativa ----------------------------
ordenes_sarima, ordenes_sarimax, filas = {}, {}, []
for sku in skus:
    y = train[sku]
    x = promo_train[[sku]].values
    ordenes_sarima[sku] = elegir_orden(y)
    ordenes_sarimax[sku] = elegir_orden(y, exog=x)

    _, m = pron_sarima(y, H, ordenes_sarimax[sku], promo_hist=x, promo_fut=promo_test[[sku]].values)
    coef, pval = m.params[m.param_names.index("x1")], m.pvalues[m.param_names.index("x1")]
    filas.append([sku, ordenes_sarima[sku], ordenes_sarimax[sku], coef, pval])

regresor = pd.DataFrame(filas, columns=["sku", "orden SARIMA (p,q,tend)", "orden SARIMAX", "efecto_promo_cajas", "p_valor"])
regresor["promo_significativa"] = regresor["p_valor"] < 0.05
regresor.round(3)


#### CELDA 7: los 4 modelos sobre el tramo de prueba (solo para mirar, NO para elegir) -----------
# La elección formal se hace con origen deslizante (punto 4). Aquí solo se verifica que todo funcione.
# SUPUESTO para SARIMAX: el calendario de promociones de las próximas 13 semanas es conocido
# (lo decide Comercial con anticipación). Si no lo fuera, se usaría promo = 0 en el futuro.
pron_test = {}
for sku in skus:
    y = train[sku]
    pron_test[(sku, "Ingenuo estacional")] = pron_naive_est(y, H)
    pron_test[(sku, "ETS")] = pron_ets(y, H)
    pron_test[(sku, "SARIMA")] = pron_sarima(y, H, ordenes_sarima[sku])[0]
    pron_test[(sku, "SARIMAX")] = pron_sarima(y, H, ordenes_sarimax[sku],
                                              promo_hist=promo_train[[sku]].values,
                                              promo_fut=promo_test[[sku]].values)[0]

modelos = ["Ingenuo estacional", "ETS", "SARIMA", "SARIMAX"]
wape_test = pd.DataFrame(index=skus, columns=modelos, dtype=float)
for (sku, mod), f in pron_test.items():
    wape_test.loc[sku, mod] = np.abs(test[sku].values - f).sum() / test[sku].sum() * 100
wape_test.loc["Promedio 9 SKU"] = wape_test.mean()
wape_test.round(1)                             # WAPE % en la prueba (menor es mejor)


#### CELDA 8: gráfico de los pronósticos en la prueba --------------------------------------------
fig, ejes = plt.subplots(3, 3, figsize=(15, 9))
for ax, sku in zip(ejes.flatten(), skus):
    ax.plot(demanda_estimada.loc[105:, sku], color="gray", label="demanda")
    for mod in modelos:
        ax.plot(test.index, pron_test[(sku, mod)], linewidth=1.2, label=mod)
    ax.axvline(FIN_TRAIN + 0.5, color="black", linestyle=":")
    ax.set_title(sku)
ejes[0, 0].legend(fontsize=7)
fig.supxlabel("Semana (línea punteada = inicio de la prueba)")
fig.supylabel("Cajas")
plt.tight_layout()
plt.show()
```

```python

from statsforecast import StatsForecast
from statsforecast.models import AutoETS, SeasonalNaive, AutoARIMA

# 1. Definir modelos (ciclo estacional de 1 año = 52 semanas)
modelos = [
    SeasonalNaive(season_length=52), # Benchmark obligatorio para el MASE
    AutoETS(season_length=52),
    AutoARIMA(season_length=52)
]

sf = StatsForecast(models=modelos, freq='W')

print("Ejecutando validación por origen deslizante... (puede tomar unos minutos)")

# 2. Validación cruzada (origen deslizante)
# h=13 y n_windows=8 según lo requerido en la Parte 4 del Avance 1
cv = sf.cross_validation(
    df=df_statsforecast,
    h=13,
    step_size=1,
    n_windows=8
)

# 3. Extraer el horizonte 'h' exacto para cada fila (Lámina 220)
dh = cv['ds'].dt.to_period('W') - cv['cutoff'].dt.to_period('W')
cv['h'] = dh.apply(lambda x: x.n)

# 4. Funciones de las métricas (Láminas 199, 205 y 220)
def wape(y, yhat): return 100 * np.sum(np.abs(y - yhat)) / np.sum(np.abs(y))
def sesgo(y, yhat): return np.mean(yhat - y) # >0 subestima, <0 sobreestima
def mae(y, yhat): return np.mean(np.abs(y - yhat))

# 5. Calcular WAPE, MASE y Sesgo desagregados por horizonte
resultados = []

for modelo in ['AutoETS', 'AutoARIMA']:
    for h_val, grupo in cv.groupby('h'):
        y_real = grupo['y']
        y_pred = grupo[modelo]
        y_naive = grupo['SeasonalNaive']

        val_wape = wape(y_real, y_pred)
        val_sesgo = sesgo(y_real, y_pred)

        # MASE = MAE del modelo / MAE del ingenuo
        mae_mod = mae(y_real, y_pred)
        mae_nai = mae(y_real, y_naive)
        val_mase = mae_mod / mae_nai if mae_nai != 0 else np.nan

        resultados.append({
            'Modelo': modelo,
            'Horizonte h': h_val,
            'WAPE (%)': val_wape,
            'MASE': val_mase,
            'Sesgo': val_sesgo
        })

# 6. Mostrar tabla final reporte_pivot
reporte_pivot = pd.DataFrame(resultados).pivot(
    index='Horizonte h',
    columns='Modelo',
    values=['WAPE (%)', 'MASE', 'Sesgo']
).round(2)

print("\n--- Reporte Desagregado por Horizonte (h) ---")
display(reporte_pivot)
```
